```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

np.random.seed(42)
random.seed(42)
```

```python
modules = ['Login', 'Payments', 'Search', 'Reports', 'User Profile']
release_types = ['Major', 'Minor', 'Hotfix']

start_date = datetime(2025, 1, 1)
num_releases = 60

release_rows = []
for i in range(1, num_releases + 1):
    release_id = i
    release_date = start_date + timedelta(days=random.randint(0, 364))
    module = random.choice(modules)
    release_type = random.choices(release_types, weights=[20, 50, 30])[0]
    release_rows.append([release_id, release_date.date(), module, release_type])

releases = pd.DataFrame(release_rows, columns=['release_id', 'release_date', 'module', 'release_type'])
releases.head(10)
```

```python
test_types = ['Functional', 'Regression', 'UAT', 'Smoke']
environments = ['Dev', 'QA', 'Staging', 'Prod']

test_case_rows = []
test_case_id = 1

for release_id in releases['release_id']:
    num_tests = random.randint(15, 40)
    for _ in range(num_tests):
        test_type = random.choices(test_types, weights=[40, 30, 15, 15])[0]
        environment = random.choices(environments, weights=[10, 40, 30, 20])[0]
        result = random.choices(['Pass', 'Fail'], weights=[85, 15])[0]
        test_case_rows.append([test_case_id, release_id, test_type, environment, result])
        test_case_id += 1

test_cases = pd.DataFrame(test_case_rows, columns=['test_case_id', 'release_id', 'test_type', 'environment', 'result'])
test_cases.head(10)
```

```python
len(test_cases)
```

```python
severities = ['Critical', 'High', 'Medium', 'Low']
found_in_options = ['QA', 'UAT', 'Production']

defect_rows = []
defect_id = 1

for release_id in releases['release_id']:
    release_row = releases[releases['release_id'] == release_id].iloc[0]
    module = release_row['module']
    release_date = release_row['release_date']

    # Login and Payments are riskier modules -> more defects
    if module in ['Login', 'Payments']:
        num_defects = random.randint(3, 12)
    else:
        num_defects = random.randint(1, 6)

    for _ in range(num_defects):
        severity = random.choices(severities, weights=[10, 25, 40, 25])[0]
        found_in = random.choices(found_in_options, weights=[50, 30, 20])[0]

        date_reported = release_date + timedelta(days=random.randint(0, 30))

        if severity == 'Critical':
            resolve_days = random.randint(1, 3)
        elif severity == 'High':
            resolve_days = random.randint(2, 7)
        elif severity == 'Medium':
            resolve_days = random.randint(5, 14)
        else:
            resolve_days = random.randint(7, 21)

        status = random.choices(['Resolved', 'Open'], weights=[80, 20])[0]
        date_resolved = date_reported + timedelta(days=resolve_days) if status == 'Resolved' else None

        defect_rows.append([defect_id, release_id, module, severity, found_in, date_reported, date_resolved, status])
        defect_id += 1

defects = pd.DataFrame(defect_rows, columns=['defect_id', 'release_id', 'module', 'severity', 'found_in', 'date_reported', 'date_resolved', 'status'])
defects.head(10)
```

```python
releases.to_csv('releases.csv', index=False)
test_cases.to_csv('test_cases.csv', index=False)
defects.to_csv('defects.csv', index=False)
print("Saved")
```

```python
from sqlalchemy import create_engine
import urllib

server = 'DESKTOP-DA7RF3C'
database = 'QualityAnalytics'

params = urllib.parse.quote_plus(
    f"DRIVER={{ODBC Driver 18 for SQL Server}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    f"Trusted_Connection=yes;"
    f"Encrypt=yes;"
    f"TrustServerCertificate=yes;"
)

engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

# quick test
with engine.connect() as conn:
    print("Connected successfully")
```

```python
releases.to_sql('releases', engine, if_exists='replace', index=False)
test_cases.to_sql('test_cases', engine, if_exists='replace', index=False)
defects.to_sql('defects', engine, if_exists='replace', index=False)

print("All tables pushed to SQL Server")
```

```python
releases_db = pd.read_sql("SELECT * FROM releases", engine)
test_cases_db = pd.read_sql("SELECT * FROM test_cases", engine)
defects_db = pd.read_sql("SELECT * FROM defects", engine)

print(releases_db.shape)
print(test_cases_db.shape)
print(defects_db.shape)
```

```python
defects_by_module = defects_db.groupby('module').size().reset_index(name='total_defects')
defects_by_module = defects_by_module.sort_values('total_defects', ascending=False)
defects_by_module
```

```python
leakage = defects_db.groupby('module').apply(
    lambda g: pd.Series({
        'total_defects': len(g),
        'production_defects': (g['found_in'] == 'Production').sum(),
        'production_leakage_pct': round((g['found_in'] == 'Production').sum() / len(g) * 100, 1)
    })
).reset_index()

leakage = leakage.sort_values('production_leakage_pct', ascending=False)
leakage
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(leakage['module'], leakage['production_leakage_pct'], color='#d62728')
plt.title('Production Defect Leakage Rate by Module')
plt.xlabel('Module')
plt.ylabel('% of Defects Reaching Production')
plt.axhline(y=leakage['production_leakage_pct'].mean(), color='gray', linestyle='--', label='Average')
plt.legend()
plt.tight_layout()
plt.savefig('production_leakage_chart.png', dpi=150)
plt.show()
```
